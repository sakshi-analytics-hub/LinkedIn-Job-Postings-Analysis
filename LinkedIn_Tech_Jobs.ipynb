```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Applicants Distribution

This graph shows how job applications are spread out, helping you see whether most jobs receive fewer or more applicants.


```python
plt.figure(figsize=(6,5))
sns.histplot(df['Total_applicants'], bins=30)
plt.title("Distribution of Applicants")
plt.show()
```

Job Level Count

```python
df["Level"].unique()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x='Level', data=df)
plt.title("Job Level Distribution")
plt.show()
```

Top Locations

```python
plt.figure(figsize=(4,3))
df['Location'].value_counts().head(10).plot(kind='bar')
plt.title("Top Locations Hiring")
plt.show()
```

Industry Distribution

```python
plt.figure(figsize=(4,3))
df['Industry'].value_counts().head(10).plot(kind='bar')
plt.title("Top Industries")
plt.show()
```

Most Demanded Skills

```python
plt.figure(figsize=(4,3))
skills = ['PYTHON','JAVA','SQL','EXCEL','REACT','AI','ML']
df[skills].sum().sort_values().plot(kind='barh')
plt.title("Top Skills Demand")
plt.show()
```

Number of Skills per Job

```python
plt.figure(figsize=(4,3))
df['skill_count'] = df[skills].sum(axis=1)
sns.histplot(df['skill_count'])
plt.title("Skill Count per Job")
plt.show()
```

Followers vs Applicants

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='LinkedIn_Followers', y='Total_applicants', data=df)
plt.title("Followers vs Applicants")
plt.show()
```

Level vs Applicants

```python
plt.figure(figsize=(4,3))
sns.boxplot(x='Level', y='Total_applicants', data=df)
plt.title("Level vs Applicants")
plt.show()
```

Top 10 Companies Hiring Most

```python
top_companies = df['Company_Name'].value_counts().head(10)

plt.figure(figsize=(5,4))
top_companies.plot(kind='bar')
plt.title("Top Companies Hiring")
plt.xticks(rotation=60)
plt.show()
```

Applicants per Company (Demand vs Competition)

```python
top_app = df.groupby('Company_Name')['Total_applicants'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(6,5))
top_app.plot(kind='bar')
plt.title("Avg Applicants per Company")
plt.xticks(rotation=60)
plt.show()
```

Designation vs Applicants

```python
top_roles = df.groupby('Designation')['Total_applicants'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,4))
top_roles.plot(kind='bar')
plt.title("Applicants by Job Role")
plt.xticks(rotation=60)
plt.show()
```

Employee Count Distribution

```python
plt.figure(figsize=(5,4))
sns.histplot(df['Employee_count'], bins=30)
plt.title("Company Size Distribution")
plt.show()
```

Company Size vs Job Count

```python
size_jobs = df.groupby('Employee_count').size()

plt.figure(figsize=(5,4))
size_jobs.plot(kind='line')
plt.title("Company Size vs Job Openings")
plt.show()
```

Industry vs Applicants

```python
industry_app = df.groupby('Industry')['Total_applicants'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,4))
industry_app.plot(kind='bar')
plt.title("Applicants by Industry")
plt.xticks(rotation=45)
plt.show()
```

Heatmap of All Skills

```python
skill_cols = df.columns[11:]  # adjust if needed

plt.figure(figsize=(8,6))
sns.heatmap(df[skill_cols].corr(), cmap='coolwarm')
plt.title("Skill Correlation Heatmap")
plt.show()
```

Location vs Applicants

```python
loc_app = df.groupby('Location')['Total_applicants'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,4))
loc_app.plot(kind='bar')
plt.title("Applicants by Location")
plt.xticks(rotation=45)
plt.show()
```

Experience Level vs Skill Requirement

```python
level_skill = df.groupby('Level')[['PYTHON','SQL','ML','AI']].sum()

level_skill.plot(kind='bar', figsize=(6,4))
plt.title("Skills by Job Level")
plt.show()
```
